이제 한 번 대충 그림을 그려 봅니다.

````
cnts['clip_enrichment'] = cnts['CLIP-35L33G.bam'] / cnts['RNA-control.bam']
cnts['rden_change'] = (cnts['RPF-siLin28a.bam'] / cnts['RNA-siLin28a.bam']) / (cnts['RPF-siLuc.bam'] / cnts['RNA-siLuc.bam'])
cnts.head()
````

````
from matplotlib import pyplot as plt
import numpy as np

fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.scatter(np.log2(cnts['clip_enrichment']),
           np.log2(cnts['rden_change']))
````

이 그림에는 문제가 많이 있습니다. 논문의 그림 Figure 4D처럼 한 번 만들어 봅시다~

````
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt

# 1. 0 이하 값 처리 (log 변환 시 -inf 발생 방지)
cnts_clean = cnts.copy()
cnts_clean.loc[cnts_clean['clip_enrichment'] <= 0, 'clip_enrichment'] = np.nan
cnts_clean.loc[cnts_clean['rden_change'] <= 0, 'rden_change'] = np.nan

# 2. log2 변환
clip_log2 = np.log2(cnts_clean['clip_enrichment'])
rden_log2 = np.log2(cnts_clean['rden_change'])

# 3. 🌟 중앙값(Median) 계산 및 중심 이동(Centering) 🌟
clip_log2_centered = clip_log2 - clip_log2.median()
rden_log2_centered = rden_log2 - rden_log2.median()

# 4. 시각화 (논문 그림 스타일 적용)
fig, ax = plt.subplots(1, 1, figsize=(5, 5))

# 산점도 그리기
ax.scatter(clip_log2_centered,
           rden_log2_centered,
           s=1,           # 점 크기
           color='black', # 점 색상
           alpha=0.3,     # 투명도 (밀집도 표현)
           zorder=2)      # 점들을 격자 위로 올라오게 설정

# 축 범위 및 눈금 설정
ax.set_xlim(-6, 4)
ax.set_ylim(-2, 2)
ax.set_xticks(np.arange(-6, 5, 2))
ax.set_yticks(np.arange(-2, 3, 1))

# 격자 설정 (논문과 동일하게 실선 형태의 회색 격자)
ax.grid(True, linestyle='-', color='lightgray', zorder=0)

# 축 라벨 및 타이틀 설정
ax.set_xlabel('LIN28A CLIP enrichment ($log_2$)')
ax.set_ylabel('Ribosome density change\nupon $Lin28a$ knockdown ($log_2$)')
ax.set_title('D  CLIP and ribosome footprinting upon\n$Lin28a$ knockdown', loc='left')

plt.tight_layout()
plt.show()
````

## Protein localization 반영하기

논문 그림(Figure 5B, S6A)에서는 protein localization을 반영해서 색으로 나타냈습니다. 이것을 한 번 구현해 봅시다. 각 gene product의 localization을 따로 구하려면 좀 귀찮기 때문에 미리 모아 두었습니다. UniProt에서 매핑된 데이터를 토대로 아주 명확한 것만 일부 정리한 것입니다. 이것도 마찬가지로 처음엔 에러가 나기도 하니까, 다시 실행해보면 거의 잘 됩니다~

````
import ssl
ssl._create_default_https_context = ssl._create_unverified_context
mouselocal = pd.read_csv('https://hyeshik.qbio.io/binfo/mouselocalization-20210507.txt', sep='\t')
mouselocal.head()
````

이제 이 localization 데이터와 위에서 만든 scatter를 결합해서 논문 그림과 비슷하게 만들어 봅시다.

````
import pandas as pd

# 1. cnts의 인덱스 초기화 (유전자 ID를 컬럼으로 빼내기)
cnts_reset = cnts.reset_index()

# 2. 기준 컬럼명이 서로 다를 경우 (cnts는 'Geneid', mouselocal은 'gene_id'라고 가정)
# 만약 위에서 확인한 이름이 다르다면 아래 'Geneid'와 'gene_id' 부분을 실제 이름으로 바꿔주세요.
merged_data = pd.merge(cnts_reset, mouselocal, left_on='Geneid', right_on='gene_id', how='left')

# 3. 결과 확인
merged_data.head()
````

````
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt

# 이전 단계에서 만들어진 merged_data_clean을 사용합니다.
# (이미 clip_log2, rden_log2 계산과 NaN 제거가 완료된 상태라고 가정합니다.)

# ---------------------------------------------------
# 🌟 핵심 보정: Median Centering (중심 맞추기) 🌟
# ---------------------------------------------------
# 전체 데이터의 X축(clip_log2)과 Y축(rden_log2)의 중앙값을 계산합니다.
median_clip = merged_data_clean['clip_log2'].median()
median_rden = merged_data_clean['rden_log2'].median()

print(f"보정 전 중앙값: CLIP={median_clip:.2f}, Rden={median_rden:.2f}")

# 각 데이터 포인트에서 중앙값을 빼주어, 새로운 중앙값이 (0,0)이 되도록 이동시킵니다.
merged_data_clean['clip_log2_centered'] = merged_data_clean['clip_log2'] - median_clip
merged_data_clean['rden_log2_centered'] = merged_data_clean['rden_log2'] - median_rden

print("✅ 보정 후 데이터의 중앙값이 (0,0)으로 이동되었습니다.")


# ---------------------------------------------------
# 보정된 데이터로 시각화
# ---------------------------------------------------
fig, ax = plt.subplots(1, 1, figsize=(5, 5))

# Cytoplasm (초록)
mask_cyto = merged_data_clean['type'] == 'cytoplasm'
ax.scatter(merged_data_clean.loc[mask_cyto, 'clip_log2_centered'], # <= 보정된 컬럼 사용!
           merged_data_clean.loc[mask_cyto, 'rden_log2_centered'], # <= 보정된 컬럼 사용!
           s=1, color='green', alpha=0.5, label='Cytoplasm', zorder=1)

# Nucleus (파랑)
mask_nuc = merged_data_clean['type'] == 'nucleus'
ax.scatter(merged_data_clean.loc[mask_nuc, 'clip_log2_centered'],
           merged_data_clean.loc[mask_nuc, 'rden_log2_centered'],
           s=1, color='blue', alpha=0.5, label='Nucleus', zorder=2)

# Integral membrane (빨강)
mask_mem = merged_data_clean['type'] == 'integral membrane'
ax.scatter(merged_data_clean.loc[mask_mem, 'clip_log2_centered'],
           merged_data_clean.loc[mask_mem, 'rden_log2_centered'],
           s=1, color='red', alpha=0.5, label='Integral membrane', zorder=3)

# 축 및 디테일 설정
ax.set_xlim(-6, 4)
ax.set_xticks(np.arange(-6, 5, 2))
ax.set_ylim(-2, 3)
ax.set_yticks(np.arange(-2, 4, 1))

ax.set_xlabel('LIN28A CLIP enrichment ($log_2$)')
ax.set_ylabel('Ribosome density change\nupon $Lin28a$ knockdown ($log_2$)')
ax.set_title('B  Protein localization', loc='left', fontsize=12, fontweight='bold')

# 중심을 더 명확히 보기 위한 십자선 추가
ax.axhline(0, color='black', linewidth=0.5, zorder=0)
ax.axvline(0, color='black', linewidth=0.5, zorder=0)

ax.yaxis.grid(True, linestyle='-', color='lightgray', zorder=0)
ax.legend(loc='upper left', scatterpoints=1, frameon=True, fontsize=9, facecolor='white', edgecolor='black', title='')

plt.tight_layout()
plt.show()
````
